VIsualization using matplotlib & seaborn

1: Churn Distribution

```python
plt.figure(figsize=(12,7))
sns.countplot(x='Churn', data=dt,palette="dark")
plt.xlabel("churn of customers",fontsize=14, fontweight='bold')
plt.ylabel("count of customers",fontsize=14, fontweight='bold')
plt.title("count of churn",fontsize=20, fontweight='bold')
plt.grid(axis='y', alpha=0.3)
plt.savefig('churn_distribution.png', dpi=200)
plt.show()
```

2: Contract vs Churn

```python
plt.figure(figsize=(10,6))

sns.countplot(x='Contract', hue='Churn', data=dt,palette="dark")
plt.title('Churn based on Contract Type',fontsize=20,fontweight='bold')
plt.xlabel('Contract Type',fontsize=14, fontweight='bold')
plt.ylabel('Number of Customers',fontsize=14, fontweight='bold')
plt.legend(title='Churn', labels=['No (Stayed)', 'Yes (Left)'],fontsize=14)
plt.grid(axis='y', alpha=0.3)
plt.savefig('churn_by_contract.png', dpi=200)
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.countplot(data=dt,x="Internetservice",hue="Churn",palette="dark")
plt.xlabel("Internetservices",fontsize=14,fontweight='bold')
plt.ylabel("Count of customers",fontsize=14,fontweight='bold')
plt.title('Internet service vs churn',fontsize=20,fontweight='bold')
plt.grid(axis='y', alpha=0.3)
plt.savefig('churn_by_internetservice.png', dpi=200)
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.boxplot(x='Churn', y='Monthlycharges', data=dt, palette='dark')
plt.title('Monthly Charges vs Churn', fontsize=20,fontweight='bold')
plt.xlabel('Churn', fontsize=14,fontweight='bold')
plt.ylabel('Monthly Charges ($)', fontsize=14,fontweight='bold')
plt.tight_layout()
plt.grid(axis='y', alpha=0.3)
plt.savefig('monthly_charges_churn.png', dpi=150)
plt.show()
plt.close()
```

```python
plt.figure(figsize=(12, 7))
sns.histplot(data=dt, x='Tenure', hue='Churn', bins=30, palette='dark',multiple='dodge')
plt.title('Tenure vs Churn', fontsize=20, fontweight='bold')
plt.xlabel('Tenure (Months)', fontsize=14, fontweight='bold')
plt.ylabel('Count of Customers', fontsize=14, fontweight='bold')
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.savefig('Tenure_vs_churn.png', dpi=200)
plt.show()
plt.close()
```

Payment_method vs Churn

```python
plt.figure(figsize=(12,7))
corr = dt.corr(numeric_only=True)
sns.heatmap(corr, annot=True,cmap='coolwarm', fmt='.2f', linewidths=0.5, linecolor='white')
plt.title("Correlation Heatmap",fontsize=20,fontweight='bold')
plt.tight_layout()
plt.savefig('correlation_heatmap.png', dpi=200)
plt.show()
```
